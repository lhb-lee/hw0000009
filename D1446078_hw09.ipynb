測試一

```python
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}
```

```python
# 1. 讀取兩個活動名稱
board_game_name, movie_name = activity_names

# 2. 整理桌遊報名名單
# 移除前後空白、轉為小寫，再以逗號分割，最後去除每個名字多餘的半形空白
board_list = [name.strip() for name in board_raw.strip().lower().split(',') if name.strip()]

# 3. 整理電影報名名單
movie_list = [name.strip() for name in movie_raw.strip().lower().split(',') if name.strip()]

# 顯示整理後的結果
print(f"活動 1: {board_game_name}")
print(f"原始桌遊名單: {board_list}")
print(f"\n活動 2: {movie_name}")
print(f"原始電影名單: {movie_list}")
```

```python
# 1. 將兩份整理後的串列轉成集合 (set)
board_set = set(board_list)
movie_set = set(movie_list)

# 2. 進行集合運算
common_participants = board_set.intersection(movie_set)       # 共同參加者
all_participants = board_set.union(movie_set)                # 全部參加者
only_board_participants = board_set.difference(movie_set)    # 只參加桌遊者
only_movie_participants = movie_set.difference(board_set)    # 只參加電影者

# 3. 輸出結果
print(f"共同參加的成員: {common_participants}")
print(f"全部參加的成員: {all_participants}")
print(f"只參加桌遊的成員: {only_board_participants}")
print(f"只參加電影的成員: {only_movie_participants}")
```

```python
# 1. 將全部參加者以字母排序並轉換為 list
sorted_participants = sorted(all_participants)

# 2. 使用切片取得前兩位作為預覽
preview_list = sorted_participants[:2]

# 3. 輸出結果
print(f"排序後的完整名單: {sorted_participants}")
print(f"通知名單預覽 (前兩位): {preview_list}")
```

```python
# 1. 計算各項指標
board_count = len(board_set)
movie_count = len(movie_set)
common_count = len(common_participants)
unique_count = len(all_participants)
total_registrations = len(board_list) + len(movie_list) # 報名人次（包含重複）

# 2. 計算應收費用 (修正為：各活動不重複人數 × 該活動單價)
board_fee_total = board_count * unit_fees[board_game_name]
movie_fee_total = movie_count * unit_fees[movie_name]
grand_total_fee = board_fee_total + movie_fee_total

# 3. 建立 summary 字典
summary = {
    "桌遊人數": board_count,
    "電影人數": movie_count,
    "共同人數": common_count,
    "總不重複人數": unique_count,
    "報名人次": total_registrations,
    "桌遊應收": board_fee_total,
    "電影應收": movie_fee_total,
    "總應收": grand_total_fee
}

# 顯示 summary 字典內容
import pprint
pprint.pprint(summary)
```

測試二

```python
# 1. 測試二原始資料（加入 Stanley 報名電影欣賞）
board_raw_2 = " Amy, Ben, amy, Cora "
movie_raw_2 = " ben, Dora, ERIC, ben, stanley "

# 2. 整理名單，同時排除重複報名的記錄（利用 dict.fromkeys 保留首次出現順序）
board_list_2 = list(dict.fromkeys([name.strip().lower() for name in board_raw_2.strip().split(',') if name.strip()]))
movie_list_2 = list(dict.fromkeys([name.strip().lower() for name in movie_raw_2.strip().split(',') if name.strip()]))

# 3. 轉成集合進行運算
board_set_2 = set(board_list_2)
movie_set_2 = set(movie_list_2)

common_participants_2 = board_set_2.intersection(movie_set_2)       # 共同參加者
all_participants_2 = board_set_2.union(movie_set_2)                # 全部不重複參加者
only_board_participants_2 = board_set_2.difference(movie_set_2)    # 只參加桌遊者
only_movie_participants_2 = movie_set_2.difference(board_set_2)    # 只參加電影者

# 4. 排序並預覽前兩位
sorted_participants_2 = sorted(all_participants_2)
preview_list_2 = sorted_participants_2[:2]

# 5. 計算各項統計指標與費用
board_count_2 = len(board_set_2)
movie_count_2 = len(movie_set_2)
common_count_2 = len(common_participants_2)
unique_count_2 = len(all_participants_2)
total_registrations_2 = len(board_list_2) + len(movie_list_2)       # 已排除重複的總報名人次 (為 7)

board_fee_total_2 = board_count_2 * unit_fees[board_game_name]
movie_fee_total_2 = movie_count_2 * unit_fees[movie_name]
grand_total_fee_2 = board_fee_total_2 + movie_fee_total_2

# 6. 建立 summary 字典
summary_2 = {
    "桌遊人數": board_count_2,
    "電影人數": movie_count_2,
    "共同人數": common_count_2,
    "總不重複人數": unique_count_2,
    "報名人次": total_registrations_2,
    "桌遊應收": board_fee_total_2,
    "電影應收": movie_fee_total_2,
    "總應收": grand_total_fee_2
}

# 7. 依照測試一的排版格式與順序輸出
print("=== 測試二：活動基本資訊 ===")
print(f"{board_game_name}: 每人 {unit_fees[board_game_name]:.2f} 元")
print(f"{movie_name}: 每人 {unit_fees[movie_name]:.2f} 元\n")

print("=== 測試二：原始報名名單 (已排除重複報名) ===")
print(f"{board_game_name} 名單: {board_list_2}")
print(f"{movie_name} 名單: {movie_list_2}\n")

print("=== 測試二：各場不重複名單 (已排序) ===")
print(f"{board_game_name} ({board_count_2} 人): {sorted(board_set_2)}")
print(f"{movie_name} ({movie_count_2} 人): {sorted(movie_set_2)}\n")

print("=== 測試二：分類成員名單 ===")
print(f"共同參加者: {sorted(common_participants_2)}")
print(f"只參加{board_game_name}者: {sorted(only_board_participants_2)}")
print(f"只參加{movie_name}者: {sorted(only_movie_participants_2)}\n")

print("=== 測試二：整體報名狀況 ===")
print(f"全部不重複參加者: {sorted_participants_2}")
print(f"總不重複人數: {unique_count_2} 人")
print(f"總報名人次 (排除重複): {total_registrations_2} 人次")
print(f"通知名單預覽 (前兩位): {preview_list_2}\n")

print("=== 測試二：應收費用統計 ===")
print(f"{board_game_name} 應收金額: {board_fee_total_2:.2f} 元")
print(f"{movie_name} 應收金額: {movie_fee_total_2:.2f} 元")
print(f"總應收金額: {grand_total_fee_2:.2f} 元\n")

print("=== 測試二：Summary 統計字典 ===")
import pprint
pprint.pprint(summary_2)
```
